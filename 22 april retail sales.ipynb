```python
date_column=pd.DatetimeIndex(rs['Date'])
rs['Year']=date_column.year
rs['Month']=date_column.month
rs['Day']=date_column.day
```

```python
display(rs)
```

```python
#sta
rs.groupby('Month')['Total Amount'].sum()
```

```python
# Sales by Category (Matplotlib)
sales_by_category = rs.groupby('Product Category')['Total Amount'].sum()
plt.figure()
sales_by_category.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
# Seaborn - Sales by Region
plt.figure()
sns.barplot(data=rs, x='Month', y='Total Amount')
plt.title('Sales by Month')
plt.show()
```

```python
#Chart shows the frequency of the sales that were made
#people buying cheaper things than the expensive things
plt.figure() #sheet of paper 
plt.hist(rs['Total Amount'], bins=10) #type of graph
plt.title('Distribution of Total Amount')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency')
plt.show()
```

```python
#Pie chart shows the percentages of sales that were made in each category
#people are buying electronics more, followed by clothing than beauty products
sales_by_category = rs.groupby('Product Category')['Total Amount'].sum()

plt.figure()
sales_by_category.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales Contribution by Category')
plt.ylabel('')
plt.show()
```

```python
#Graph shows us sales that were made from the beginning of 2023 to 2024 (per day)
sales_trend = rs.groupby('Date')['Total Amount'].sum()
plt.figure(figsize=(10,5))# 10(is the sides) and 5(is the height)
plt.plot(sales_trend, marker='o')
plt.title('Sales Trend with Markers')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```

```python
# Sales Trend Over Time
rs['Date'] = pd.to_datetime(rs['Date'])
sales_trend = rs.groupby('Date')['Total Amount'].sum()
plt.figure(figsize=(12,5)) #12(drag sides) 5(height)
sales_trend.plot(color="red") #changing color 
plt.title('Sales Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```
